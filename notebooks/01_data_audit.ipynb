# 01 — Data Audit

Inspect the Kaggle training data, data types, missing values, duplicates, and sales statistics.

```
from pathlib import Path
import pandas as pd

PROJECT = Path("/content/drive/MyDrive/retail-demand-forecasting")
DATA_DIR = PROJECT / "data/raw"
train = pd.read_csv(DATA_DIR / "train.csv", parse_dates=["date"])

print("Dataset shape:", train.shape)
print("Date range:", train["date"].min(), "to", train["date"].max())
print("Unique stores:", train["store_nbr"].nunique())
print("Product families:", train["family"].nunique())
print("Duplicate rows:", train.duplicated().sum())
print("\nMissing values:")
display(train.isna().sum().to_frame("missing_count"))
print("\nSales statistics:")
display(train["sales"].describe().to_frame("sales"))
print("\nFirst five rows:")
display(train.head())
```
